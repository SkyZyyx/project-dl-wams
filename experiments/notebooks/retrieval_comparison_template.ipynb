# Retrieval Model Comparison

This notebook reads the CSV artifacts generated by `experiments.retrieval.cli compare` and helps compare CLIP against the DINOv2 variants.

```python
from pathlib import Path
import pandas as pd

root = Path('../artifacts/runs').resolve()
summary = pd.read_csv(root / 'comparison_summary.csv')
summary
```

```python
columns = [
    'model_name',
    'recall_at_10_product',
    'map_at_10_product',
    'ndcg_at_10_product',
    'embed_latency_ms',
]
summary[columns].sort_values('recall_at_10_product', ascending=False)
```

```python
ax = summary.plot.bar(x='model_name', y=['recall_at_1_product', 'recall_at_5_product', 'recall_at_10_product'], figsize=(12, 6))
ax.set_ylabel('Recall')
ax.set_title('Product-level Recall@K by Model')
```
